The AVG column contains "-" for some players. These players were never dismissed,
so their batting average cannot be calculated using Runs / (Innings - Not Outs).
We will convert "-" into a missing value (NaN).

```python
df['AVG'] = pd.to_numeric(df['AVG'], errors='coerce')
df['AVG'].head(10)
```

```python
df['AVG'].isna().sum()
```

```python
df['HS_clean'] = df['HS'].str.replace('*', '', regex=False).astype(int)
df[['HS', 'HS_clean']].head(10)
```

```python
df.info()
```

**Q1- Top 10 Run Scorers**

We will sort all batters according to total runs in descending order
and select the top 10 players.

```python
top10_runs = df.sort_values('Runs', ascending=False).head(10)
top10_runs[['Player Name', 'Team', 'Runs']]
```

**Q1 Answer**

The table above shows the top 10 run-scorers of IPL 2025 along with their teams.
The highest run-scorer can be considered a high-output batting asset because
of the large number of runs contributed during the season.

**Q2- Team-Wise Total Runs**

We will group the batters according to their teams and add their total runs.

```python
team_runs = df.groupby('Team')['Runs'].sum().sort_values(ascending=False)
team_runs
```

**Q2 Answer**

The table above ranks all IPL teams according to the total runs scored
by their batters. The team at the top has the highest collective batting output.

**Q3- Fastest Scorers**

To avoid the Strike Rate illusion, we will only consider batters
who faced at least 100 balls.

```python
fastest_scorers = df[df['BF'] >= 100].sort_values(
    'SR',
    ascending=False
).head(10)
fastest_scorers[['Player Name', 'Team', 'BF', 'SR']]
```

**Q3 Answer**

After applying the minimum requirement of 100 balls faced,
the table above shows the 10 fastest scorers based on Strike Rate.
The BF filter makes the comparison more meaningful because it avoids
giving too much importance to players with very few opportunities.

**Q4- Most Reliable Batters**

We will consider only players who batted in at least 8 innings
and rank them according to batting average.

```python
reliable_batters = df[df['Inn'] >= 8].sort_values(
    'AVG',
    ascending=False
).head(10)
reliable_batters[['Player Name', 'Team', 'Inn', 'Runs', 'AVG']]
```

**Q4 Answer**

The table above shows the most reliable batters based on batting average,
considering only players who batted in at least 8 innings.
A higher average indicates greater reliability in converting batting
opportunities into runs.

**Q5- Boundry Percentage**

Boundary% measures what percentage of a batter's total runs
came directly from fours and sixes.

```python
df['Boundary%'] = ((df['4s'] * 4) + (df['6s'] * 6)) / df['Runs'] * 100
df[['Player Name', 'Runs', '4s', '6s', 'Boundary%']].head(10)
```

```python
df['Boundary%'] = (
    (df['4s'] * 4 + df['6s'] * 6)
    / df['Runs']
) * 100
top_boundary_hitters = df.sort_values(
    'Boundary%',
    ascending=False
).head(10)
top_boundary_hitters[
    ['Player Name', 'Team', 'Runs', '4s', '6s', 'Boundary%']
]
```

**Q5 Answer**

Boundary% indicates how much of a batter's total scoring came from
fours and sixes. A high Boundary% indicates strong boundary-hitting ability,
which can be valuable to a franchise because boundaries can accelerate
the scoring rate and create greater match-winning impact.

```python
# Import matplotlib for creating charts
import matplotlib.pyplot as plt

# Select the top 10 run scorers
top10 = df.sort_values('Runs', ascending=False).head(10)

# Create the bar chart
plt.figure(figsize=(12, 6))

plt.bar(top10['Player Name'], top10['Runs'])

# Add chart title
plt.title('Top 10 Run Scorers - IPL 2025')

# Add axis labels
plt.xlabel('Player')
plt.ylabel('Total Runs')

# Rotate player names so they are readable
plt.xticks(rotation=75)

# Display the chart
plt.show()
```

```python
# Q2 Visualization: Team-wise total runs

team_runs = df.groupby('Team')['Runs'].sum().sort_values(
    ascending=False
)

# Create the bar chart
plt.figure(figsize=(12, 6))

plt.bar(team_runs.index, team_runs.values)

# Add title
plt.title('Team-wise Total Runs - IPL 2025')

# Add axis labels
plt.xlabel('Team')
plt.ylabel('Total Runs')

# Rotate team names
plt.xticks(rotation=45)

# Display the chart
plt.show()
```

```python
# Select batters who faced at least 100 balls
qualified_batters = df[df['BF'] >= 100]

# Create scatter plot
plt.figure(figsize=(10, 7))

plt.scatter(
    qualified_batters['AVG'],
    qualified_batters['SR']
)

# Add title
plt.title('Batting Average vs Strike Rate - IPL 2025')

# Add axis labels
plt.xlabel('Batting Average (AVG)')
plt.ylabel('Strike Rate (SR)')
# Add grid to make the chart easier to read
plt.grid(True)
# Display the chart
plt.show()
```
